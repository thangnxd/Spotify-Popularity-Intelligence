```python
import numpy as np #Làm việc với các dữ liệu mảng nhiều chiều
import pandas as pd #Giúp làm việc với các dữ liệu dạng bảng
```

```python
high_path = "data/high_popularity_spotify_data.csv"
low_path = "data/low_popularity_spotify_data.csv"

high_df = pd.read_csv(high_path)
low_df = pd.read_csv(low_path)

print("High shape:", high_df.shape)
print("Low shape:", low_df.shape)
```

```python
high_df["popularity_group"] = "High"
low_df["popularity_group"] = "Low"

df = pd.concat([high_df, low_df], ignore_index=True)

print("Combined shape:", df.shape)
df.head()
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicated rows:", df.duplicated().sum())

df.info()
```

```python
drop_cols = [
    "track_href",
    "uri",
    "analysis_url",
    "track_id",
    "track_album_id",
    "playlist_id",
    "id",
    "type"
]

df_clean = df.drop(columns=[col for col in drop_cols if col in df.columns])

df_clean.head()
```

```python
# Xóa duplicate nếu có
df_clean = df_clean.drop_duplicates()

# Với album name bị thiếu thì điền Unknown
if "track_album_name" in df_clean.columns:
    df_clean["track_album_name"] = df_clean["track_album_name"].fillna("Unknown")

print("Final shape:", df_clean.shape)
print(df_clean.isnull().sum())
```

```python
df_clean = df_clean.dropna()
print("Final shape:", df_clean.shape)
print(df_clean.isnull().sum())
```

```python
output_path = "data/spotify_cleaned.csv"

df_clean.to_csv(output_path, index=False, encoding="utf-8-sig")

print("Saved to:", output_path)
```
